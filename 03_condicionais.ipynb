```python
# D3

vendas = [1200, -50, 0, 3400, -200, 800]
total_vendas = 0
total_estornos = 0

for v in vendas:
  if v > 0:
    total_vendas += v
  elif v < 0:
    total_estornos -= v

print(f'Total de vendas: {total_vendas}, total de estornos: {abs(total_estornos)}.')    # abs() devolve o valor absoluto

```

```python
# E1

def verificar_par_impar(numero):
  if numero % 2 == 0:
    return f"{numero} é par."
  else:
    return f"{numero} é impar."

verificar_par_impar(7)
```

```python
# E2

def calcular_imc(peso, altura):

  imc = peso/altura**2

  if imc < 18.5:
    return f'Abaixo do peso. IMC: {imc:.2f}'
  elif imc < 24.9:
    return f'Peso normal. IMC: {imc:.2f}'
  else:
    return f'Sobrepeso: IMC {imc:.2f}'

calcular_imc(72, 1.78)
```

```python
# E3

def pode_aposentar(idade, tempo_contribuicao, genero):
  if idade >= 62 and tempo_contribuicao >= 15 and genero == 'F':
    return True
  elif idade >= 65 and tempo_contribuicao >= 15 and genero == 'M':
    return True
  else:
    return False

print(pode_aposentar(62, 15, 'F'))

```
