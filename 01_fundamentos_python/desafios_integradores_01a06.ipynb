```python
# Desafio 03 - CORREÇÂO

def calcular_fatura(*valores_de_compra):
  qtde_itens = len(valores_de_compra)

  # previne o caso de divisão por 0
  if qtde_itens == 0:
    return None

  # usando a variável total_fatura no cálculo da média para evitar chamar sum() duas vezes
  total_fatura = sum(valores_de_compra)
  valor_medio_itens = total_fatura / qtde_itens

  # tratamento de valores zerados ou negativos
  if total_fatura <= 0:
    return None

  # adicionando desconto de 10% para faturas de 500 reais ou mais
  if total_fatura >= 500:
    total_fatura -= (total_fatura * 0.10)

  # O python permite retornar multiplos valores como tupla
  return total_fatura, qtde_itens, valor_medio_itens

# --- Testando e desempacotando os resultados ---

testes = [
    (100, 200, 300, 400, 500), # Mais de 500 (Desconto)
    (50, 100, 150, 200),       # Exatamente 500 (Sem desconto, pelo seu critério atual)
    (10, 20, 30, 40, 50),      # Menos de 500
    (0, 0, 0, 0),              # Zerado
    ()                         # Vazio
]
for teste in testes:
  resultado = calcular_fatura(*teste)

  if resultado is None:
    print("Fatura inválida")
  else:
    # "Desempacotando" a tupla retornada nas 3 variáveis
    total, qtde, media = resultado
    print(f"Total: R$ {total:.2f} | Itens: {qtde} | Média: R$ {media:.2f}")







```

```python
# Desafio 04

def contar_caracter(texto, caractere):

  # tratar se não for ou não tiver caractere.
  if texto == "":
    return 0

  # checa se o primeiro caractere do texto é igual o caractere parâmetro
  if texto[0] == caractere:
    # se for, retorna 1 + o resultado da função do restante do texto.
    return 1 + contar_caracter(texto[1: ], caractere)
  else:
    # se não for, retorna 0 + o resultado da função do restante do texto.
    return 0 + contar_caracter(texto[1: ], caractere)

print(contar_caracter('arara', 'a'))



```

```python
# desafio 05

vendas = [
    {"vendedor": "Carla", "produto": "Motor", "valor": 1200},
    {"vendedor": "Diego", "produto": "Painel", "valor": 800},
    {"vendedor": "Carla", "produto": "Sensor", "valor": 350},
    {"vendedor": "Diego", "produto": "Motor", "valor": 1200},
    {"vendedor": "Carla", "produto": "Painel", "valor": 800},
    {"vendedor": "Eduardo", "produto": "Sensor", "valor": 350},
]

# passo 01: somar e imprimir  o total das vendas
soma_vendas = 0
for venda in vendas:
    soma_vendas += venda['valor']

print(f'Valor total das vendas: R$ {soma_vendas:.2f}\n')

# passo 02: encontrar o top vendedor e imprimir (sem usar max()). preciso da soma por vendedor
vendas_por_vendedor = {}
for venda in vendas:
    vendas_por_vendedor[venda['vendedor']] = vendas_por_vendedor.get(venda['vendedor'], 0) + venda['valor']

# ainda no passo 02, agora que tenho a soma de vendas por vendedor consigo o top vendedor
maior_venda = 0
top_vendedor = ''
# lembra do metodo, acessando chave e valor
for vendedor, valor in vendas_por_vendedor.items():
    if valor > maior_venda:
        maior_venda = valor
        top_vendedor = vendedor

print(f'Top vendedor: {top_vendedor} | R$ {maior_venda:.2f}\n')

# passo 03: classificando vendedores através de uma função
def top_vendedores(vendas, limite_vendas):
    qualificados = []
    # aqui meu argumento é o novo dict com vendedor e total por vendedor
    for vendedor, total in vendas.items():
        if total >= limite_vendas:
            qualificados.append(vendedor)
    return qualificados

print(f'Top vendedores: {top_vendedores(vendas_por_vendedor, 1000)}')

# passo 04: criando categorias de vendedores
def classificar_vendedores(dados_vendedores):
    categorias = {}
    for vendedor, total in dados_vendedores.items():
        if total > 2000:
            categorias[vendedor] = 'Destaque'       # sem aspas para o python não sobrescrever a chave 'vendedor'
        elif total >= 1000:                         # e pegar o nome da pessoa
            categorias[vendedor] = 'Regular'
        else:
            categorias[vendedor] = 'Abaixo'
    return categorias

print(f'\nDesempenho dos vendedores: {classificar_vendedores(vendas_por_vendedor)}')



```

```python
# Desafio 06 - Mini sistema de chamados

chamados = [
    {"fazenda": "Terra Norte", "tecnico": "Rafael", "valor": 1200, "horas": 3, "status": "Concluído"},
    {"fazenda": "Boa Vista", "tecnico": "João", "valor": 850, "horas": 26, "status": "Concluído"},
    {"fazenda": "Rio Claro", "tecnico": "Rafael", "valor": 2300, "horas": 8, "status": "Concluído"},
    {"fazenda": "Terra Norte", "tecnico": "João", "valor": 450, "horas": 50, "status": "Aberto"},
    {"fazenda": "Boa Vista", "tecnico": "Rafael", "valor": 900, "horas": 1, "status": "Concluído"},
    {"fazenda": "Rio Claro", "tecnico": "João", "valor": 0, "horas": 12, "status": "Cancelado"},
]

# Passo 01: somando somente os concluídos
def faturamento_total(lista_chamados):
    soma_concluidos = 0
    # a função tem q ter o parametro como variavel interNAAAAAAAAAAAAAAAAAA e não a variável global
    # pensa def função(parametro) -- for f in *PARAMETROOOOOOOO*
    for item in lista_chamados:
        if item['status'] == 'Concluído':
            soma_concluidos += item['valor']

    return soma_concluidos

# passo 02: classificar atendimento pelo tempo
# aqui a função faz um trabalho UNICO
def classificar_atendimento(tempo_chamados):
    # condiionais para o PARAMETROOOOOOO
    if tempo_chamados < 0: # parametro
        return "Inválido"
    elif tempo_chamados < 4: # parametro
        return "Exelente"
    elif tempo_chamados < 24: # parametro
        return 'Dentro do prazo'
    else:
        return 'Atrasado'

# passo 03: chamados concluidos por tecnico
def chamados_por_tecnico(lista_chamados):
    qtde_por_tecnico = {}

    for item in lista_chamados:
        if item['status'] == 'Concluído':
            # esse ja faz o trabalho do contador
            qtde_por_tecnico[item['tecnico']] = qtde_por_tecnico.get(item['tecnico'], 0) + 1

    return qtde_por_tecnico

# Passo 04: relatório completo(composição das funções)
def relatorio_completo(lista_chamados):
    # 1. Obtém o faturamento usando a primeira função
    faturamento = faturamento_total(lista_chamados)
    # 2. Obtém o dicionário por técnico usando a terceira função
    tecnicos = chamados_por_tecnico(lista_chamados)
    # 3. Gera a lista de classificações chamando a segunda função para cada chamado Concluído

    classificados = []
    for chamado in lista_chamados:
        if chamado['status'] == 'Concluído':
            classificacao = classificar_atendimento(chamado['horas'])
            classificados.append(f"{chamado['fazenda']} : {classificacao}")

    # Retorna os 3 valores exigidos (tupla)
    return faturamento, tecnicos, classificados

# Exibindo os resultados:

faturado_total, por_tecnico, lista_classificacoes = relatorio_completo(chamados)

print(f"Faturamento Total: R$ {faturado_total:.2f}")
print(f"Chamados Concluídos por Técnico: {por_tecnico}")
print(f"Classificação dos Atendimentos Concluídos: {lista_classificacoes}")












```

```python
# Desafio 07 - conversor de temperatura

# passo 01: função que converte c para f
def celsius_para_fahrenheit(c):
    f = (c * 9/5) + 32
    return f

# passo 02: função que classifica as temperaturas convertidas
def classificar_temperatura(f):
    if f < 50:
        return 'Frio'
    elif f <= 85:
        return 'Ameno'
    else:
        return 'Quente'

temp_c = [0, 15, 28, 33, - 5, 40]
temp_f = []

# passo 03: percorrer a lista com as temperaturas em celsius, converte-las e adicionalas em uma nova lista
for c in temp_c:
    temp_farenheit = celsius_para_fahrenheit(c)
    temp_f.append(temp_farenheit)

# passo 04: percorrer a nova lista de temperaturas em farenheit e classifica-las, e imprimir a classificação
for f in temp_f:
    classificacao = classificar_temperatura(f)
    print(f'{f}°F: {classificacao}')






```

```python
# Desafio 08 - teste de palíndromo

# Passo 01: função recursiva
def _verificar_recursivo(texto):
    if len(texto) <= 1:
        return True
    elif texto[0] == texto[-1]:
       return _verificar_recursivo(texto[1:-1])
    else:
        return False

# passo 02: função auxiliar que limpa espaços, e transforma maiusculas em minusculas
def eh_palindromo(texto_bruto):
    # limpeza:  maiscusuclas e espaços
    texto_limpo = texto_bruto.lower().replace(' ', '')
    return _verificar_recursivo(texto_limpo)

print(eh_palindromo('Arara'))
print(eh_palindromo('Radar'))
print(eh_palindromo('Onibus'))
print(eh_palindromo(''))


```

```python
# desafio 13 -- Calculadora de frete por distancia

def calcular_frete(peso, distancia):

    if peso > 50 or distancia > 500:
        valor_frete = 200
    elif peso > 20:
        valor_frete = (peso * 2) + (distancia * 0.1)
    else:
        valor_frete = (peso*1.5) + (distancia*0.05)

    return valor_frete

print(f'Valor do frete: R$ {calcular_frete(55, 510):.2f}')
print(f'Valor do frete: R$ {calcular_frete(15, 490):.2f}')
print(f'Valor do frete: R$ {calcular_frete(25, 700):.2f}')
print(f'Valor do frete: R$ {calcular_frete(55, 10):.2f}')
print(f'Valor do frete: R$ {calcular_frete(0, 0):.2f}')



```

```python
# desafio 14 - removedor de dupliacadas com contagem

pedidos = [101, 105, 101, 108, 105, 101, 110]

def processar_pedidos(lista_pedidos):
    quantidade_duplicatas = {}

    # 01. Monta o dicinário de contagem
    for pedido in lista_pedidos:
        quantidade_duplicatas[pedido] = quantidade_duplicatas.get(pedido, 0) + 1

    # 02. extraindo as chaves únicas na ordem em que aparecem pela primeira vez
    pedidos_unicos = list(quantidade_duplicatas.keys())

    return pedidos_unicos, quantidade_duplicatas

unicos, contagem = processar_pedidos(pedidos)

print(f'Pedidos únicos: {unicos}')
print(f'Frequência: {contagem}')

```

```python
# desafio 15 - sequencia de collatz

def collatz(numero):
    # Já começa com o número de entrada
    sequencia_numeros = [numero]

    while numero != 1:
        if numero % 2 == 0:
            numero = numero // 2
        else:
            numero = 3 * numero + 1
        sequencia_numeros.append(numero)
    return sequencia_numeros

print(collatz(6))
```

```python
# desafio 15 -- sequecia de collatz -- recursivo

def collatz(numero):
    if numero == 1:
        return [1]
    elif numero % 2 == 0:
        return [numero] + collatz(numero // 2)
    else:
        return [numero] + collatz(3 * numero + 1)

print(collatz(6))
```

```python
# desafio 16 - Sistema de aprovação com multiplos critérios

alunos = [
    {"nome": "Ana", "frequencia": 92, "nota": 7.5},
    {"nome": "Bruno", "frequencia": 60, "nota": 8.0},
    {"nome": "Carla", "frequencia": 85, "nota": 4.5},
    {"nome": "Diego", "frequencia": 60, "nota": 5.5},
]

# função classificadora independente
def status_aluno(aluno):
    if aluno['frequencia'] < 75 and aluno['nota'] < 6:
        return 'Reprovado por falta e nota'
    elif aluno['frequencia'] < 75:
        return 'Reprovado por falta'
    elif aluno['nota'] < 6:
        return 'Reprovado por nota'
    else:
        return 'Aprovado'

# percorrendo a lista de alunos
for aluno_ in alunos:
    status = status_aluno(aluno_)
    print(f'{aluno_["nome"]}: {status}')



```

```python
# desafio 17 - simulador de investimento

def simular_investimento(capital_inicial, taxa_mensal, meses):
    saldo_por_mes = []
    saldo_atual = capital_inicial

    for mes in range(1, meses + 1):
        saldo_atual = saldo_atual * (1 + taxa_mensal)
        saldo_por_mes.append(round(saldo_atual, 2))

    return saldo_por_mes

print(simular_investimento(1000, 0.01, 1))
```

```python
# desafio 18 - filtro multicriterio de produtos

lista_produtos = [
    {"nome": "Teclado", "categoria": "Eletrônicos", "preco": 150, "estoque": 10},
    {"nome": "Monitor", "categoria": "Eletrônicos", "preco": 800, "estoque": 5},
    {"nome": "Cadeira", "categoria": "Móveis", "preco": 350, "estoque": 2},
    {"nome": "MousePad", "categoria": "Eletrônicos", "preco": 50, "estoque": 20}
]

def filtrar_produtos(produtos, categoria=None,preco_max=None):
    produtos_filtrados = []

    for produto in produtos:
        # se categoria "nao nada"(ou seja, o filtro existe) e e a chave "categoria" != do filtro
        if categoria is not None and produto['categoria'] != categoria:
            continue
        # se preco_max "nao nada"(ou seja, o filtro existe) e a chave "preco" != do filtro
        elif preco_max is not None and produto['preco'] > preco_max:
            continue
        else:
            produtos_filtrados.append(produto)

    return produtos_filtrados

print(filtrar_produtos(lista_produtos, preco_max=50))
print(filtrar_produtos(lista_produtos, categoria='Eletrônicos'))
print(filtrar_produtos(lista_produtos))
print(filtrar_produtos(lista_produtos, categoria='Móveis', preco_max=350))

```

```python
# desafio 18 - filtro multicriterio de produtos - booleanos

lista_produtos = [
    {"nome": "Teclado", "categoria": "Eletrônicos", "preco": 150, "estoque": 10},
    {"nome": "Monitor", "categoria": "Eletrônicos", "preco": 800, "estoque": 5},
    {"nome": "Cadeira", "categoria": "Móveis", "preco": 350, "estoque": 2},
    {"nome": "MousePad", "categoria": "Eletrônicos", "preco": 50, "estoque": 20}
]

def filtrar_produtos(produtos, categoria=None, preco_max=None):
    produtos_filtrados = []

    for produto in produtos:
        # 1. Valida Categoria: passa se for None OU se for igual à informada
        passou_categoria = categoria is None or produto['categoria'] == categoria
        # 2. Valida Preço: passa se for None OU se for menor/igual ao limite
        passou_preco = preco_max is None or produto['preco'] <= preco_max

        # 3. O produto só entra se passar em AMBOS os critérios
        if passou_preco and passou_categoria:
            produtos_filtrados.append(produto)

    return produtos_filtrados

print(filtrar_produtos(lista_produtos, preco_max=50))
print(filtrar_produtos(lista_produtos))
print(filtrar_produtos(lista_produtos, categoria='Móveis', preco_max=350))
print(filtrar_produtos(lista_produtos, categoria='Eletrônicos'))



```

```python
# desafio 19 - rotacao de plantao

tecnicos = ["Rafael", "João", "Marcos"]
dias = 10

def gerar_escala(lista_tecnicos, total_dias):
    escala = []
    total_tecnicos = len(lista_tecnicos)

    for dia in range(total_dias):
        indice_tecnico = dia % total_tecnicos
        tecnico_atual = lista_tecnicos[indice_tecnico]
        escala.append(f'Dia {dia + 1}: {tecnico_atual}')

    return escala

escala_plantao = gerar_escala(tecnicos, dias)
for linha in escala_plantao:
    print(linha)
```

```python
# desafio 20 - Mini Sistema de Chamados com Menu

chamados = []

while True:
    opcao_menu = input('\nEscolha uma opção:\n1. Abrir chamado\n2. Listar chamados\n3. Ver total Faturado\n4. Sair\n')

    if opcao_menu not in ('1', '2', '3', '4'):
        print('\nOpção inválida\n')
        continue

    elif opcao_menu == '1':
        adicionar_fazenda = input('\nDigite o nome da Fazenda: ')
        adicionar_tecnico = input('Digite o nome do Técnico: ')
        valor_chamado = float(input('Digite o valor do chamado: '))
        chamado = {'Fazenda': adicionar_fazenda, 'Técnico': adicionar_tecnico, 'Valor': valor_chamado}
        chamados.append(chamado)

    elif opcao_menu == '2':
        if not chamados:
            print("\nNenhum chamado registrado até o momento.")
        else:
            print("\n--- RELATÓRIO DE CHAMADOS ---")
            for i, c in enumerate(chamados, start=1):
                print(f"{i}. Fazenda: {c['Fazenda']} | Técnico: {c['Técnico']} | Valor: R$ {c['Valor']:.2f}")
            print("-" * 30)

    elif opcao_menu == '3':
        total_faturado = sum(chamado['Valor'] for chamado in chamados)
        print(f'\nTotal faturado: R$ {total_faturado:.2f}')

    elif opcao_menu == '4':
        print('\nSaindo...')
        break



```
