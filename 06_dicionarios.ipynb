```python
# C1

precos = {
    'parafuso': 0.5,
    'porcas': 0.3,
    'arruela': 0.2,
}

# .items -- todas as chaves: dict_keys(['nome', 'preco', ...])
# util para trabalhar com os pares: chave e valor.
for key, value in precos.items():
  print(f'O preço do {key} é R$ {value}')
```

```python
# C2

soma = 0
precos = {
    'parafuso': 0.5,
    'porcas': 0.3,
    'arruela': 0.2,
}

# .values -- todos os valores: dict_values(['Parafuso', 0.55, ...])
# util para trabalhar somente com os valores do dict quando a chave (ou nome) não importa
for values in precos.values():
  soma += values

print(f' R$ {soma}')
```

```python
# E2

# A função recebe o nome e a quantidade que será adicionada/removida
def atualizar_estoque(nome, quantidade):

    # Ele(o dicionário) está dentro da função porque o exercício foi estruturado para a
    # função criar e trabalhar com esse dicionário internamente.
    # O dicionário é criado dentro da função porque ela trabalha com um novo produto a cada chamada
    produto = {
        'nome': 'parafuso',
        'estoque': 10,
    }

    # Verifica antes se a alteração deixaria o estoque negativo
    if produto['estoque'] + quantidade < 0:
        return 'Estoque insuficiente'

    # Se houver estoque suficiente, realiza a alteração
    else:
        produto['estoque'] += quantidade

        # Atualiza o nome do produto
        produto['nome'] = nome

        # Retorna o dicionário já atualizado
        return produto
```

```python
# E3

# A lista existe fora da função e será passada como argumento
produtos = [
      {'nome': 'parafuso', 'preco': 0.5, 'estoque': 100},
      {'nome': 'porcas', 'preco': 0.3, 'estoque': 50},
  ]

# A função recebe uma lista e o nome que queremos procurar
def buscar_por_nome(lista_produtos, nome_procurado):

  # Percorre cada dicionário dentro da lista
  for produto in lista_produtos:

    # Verifica se o nome do produto atual é igual ao nome procurado
    if produto['nome'] == nome_procurado:

      # Se encontrar, retorna o dicionário inteiro
      return produtos

   # Se terminar o for sem encontrar, retorna None
  return None

print(buscar_por_nome(produtos, 'parafuso'))
print(buscar_por_nome(produtos, 'porcas'))
print(buscar_por_nome(produtos, 'arruela'))
```

```python
# Bonus


chamados = [
    {"fazenda": "Terra Norte", "valor": 1200, "tecnico": "Rafael", "status": "Concluído"},
    {"fazenda": "Boa Vista", "valor": 850, "tecnico": "João", "status": "Aberto"},
    {"fazenda": "Rio Claro", "valor": 2300, "tecnico": "Rafael", "status": "Concluído"},
    {"fazenda": "Terra Norte", "valor": 450, "tecnico": "João", "status": "Concluído"},
    {"fazenda": "Boa Vista", "valor": 900, "tecnico": "Rafael", "status": "Aberto"},
]

soma_concluidos = 0
chamados_por_tecnico = {}

# passo 1 - verificar quantos chamados foram concluidos e somar os valores
for i in chamados:
  if i['status'] == 'Concluído':
    soma_concluidos += i['valor']

print(f' A soma dos chamados concluídos é: R$ {soma_concluidos}')

# passo 2 - verificar quantos chamados cada tecnico atendeu(concluiu)
for i in chamados:
  if i['status'] == 'Concluído':

    # .get(chave, 0) pega o valor atual daquele técnico no dicionário, OU 0 se ele ainda não existir ali
    # depois soma +1 e guarda de volta na mesma chave -- cria a chave na primeira vez, atualiza nas seguintes
    # equivale a: if tecnico in dicionario: dicionario[tecnico] += 1 / else: dicionario[tecnico] = 1
    chamados_por_tecnico[i['tecnico']] = chamados_por_tecnico.get(i['tecnico'], 0) + 1

print(chamados_por_tecnico)




```
