```python
# B2

soma = 0

while True:    # While True + Break: Repete até uma condição específica mandar parar
  resposta = (input('Digite um numero: '))
  if resposta == 'Parar':
    break
  else:
    soma += int(resposta)

print(f'A soma dos números é: {soma}')

```

```python
# D3

senha_correta = "python123"
numero_tentativas = 3

while True:
  senha = input('Digite a senha: ')
  numero_tentativas -= 1

  if senha == senha_correta:
    print('Acesso permitido!')
    break

  if numero_tentativas == 0:
    print('Acesso Bloqueado! Você excedeu o número de tentativas.')
    break
```

```python
# E1

def numero_primo(numero):

  if numero < 2:
    return False

  else:
    for i in range(2, numero):
      if numero % i == 0:
        return False
    return True

numero_primo(13)
```

```python
# E2

def contar_vogais(palavra):
  contador_vogais = 0

  for letra in palavra:
    if letra in 'aeiouAEIOU':
      contador_vogais += 1
  return contador_vogais

print(f'A palavra tem {contar_vogais('Vasco')} vogais!')
```

```python
# E3

def fibonacci(n):
  a = [0, 1]

  while len(a) < n:
    prox = a[-1] + a[-2]
    a.append(prox)
  return a

fibonacci(10)
```

```python
# Bônus

tempos_resposta = [3, 26, 8, 50, 8, 26, 28]

atrasado_anterior = False   # guarda se o chamado da volta ANTERIOR foi "Atrasado" (começa False pois não existe anterior na 1ª volta)
i = 0                       # índice manual, já que estamos usando while em vez de for

while i < len(tempos_resposta):     # repete enquanto ainda houver posição válida na lista
  tempo = tempos_resposta[i]        # pega o valor da posição atual usando o índice

  # classifica o tempo de resposta atual
  if tempo < 4:
    classificacao = 'Excelente!'
  elif tempo < 24:
    classificacao = 'Dentro do Prazo'
  else:
    classificacao = 'Atrasado'

  print(classificacao)

  # só dispara o break se o ATUAL for Atrasado E o ANTERIOR também tiver sido
  # (ou seja: dois atrasados em sequência)
  if classificacao == 'Atrasado' and atrasado_anterior:
    print(f'Dois atrasos seguidos! Encontrados no índice {i}.')  # i ainda não foi incrementado aqui, então aponta certo pro item atual
    break

  i += 1   # avança pro próximo índice (SÓ depois de já ter usado i para tudo que precisava nesta volta)

  # atualiza a flag para a PRÓXIMA volta: vira True se o atual foi Atrasado, False caso contrário
  atrasado_anterior = (classificacao == 'Atrasado')






```
