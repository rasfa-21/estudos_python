```python
# A1. Dada a lista temperaturas_motores = [68, 75, 82, 90, 65, 95, 102, 70], 
# gere uma nova lista contendo apenas as temperaturas acima de 80°C.

temperatura_motores = [68, 75, 82, 90, 65, 95, 102, 70]
temperaturas_elevadas = [t for t in temperatura_motores if t > 80]

print(temperaturas_elevadas)
```

```python
# A2. Dada a lista de pressões hidráulicas em bar pressoes = [12.5, 14.0, 11.8, 15.2], 
# gere uma lista convertendo todos os valores para PSI (multiplique cada valor por 14.5038).

pressoes_bar =  [12.5, 14.0, 11.8, 15.2]
pressoes_psi = [p * 14.5038 for p in pressoes_bar]

print(pressoes_psi)
```

```python
#A4. Dada a lista leituras = [10, 0, 25, 0, 42, 0, 50], 
# filtre e retorne uma lista contendo apenas as leituras diferentes de zero.

leituras = [10, 0, 25, 0, 42, 0, 50]
leituras_num = [l for l in leituras if l != 0]

print(leituras_num)

```

```python
#A5. Dada a lista de horas trabalhadas por máquina horas = [120, 450, 800, 1200, 240, 1600], 
# gere uma lista apenas com as horas que ultrapassaram a janela de revisão preventiva (maior ou igual a 500 horas).

horas_por_maquina = [120, 450, 800, 1200, 240, 1600]
horas_limite = [h for h in horas_por_maquina if h >= 500]

print(horas_limite)
```

```python
# A6. Dada a lista ["  Carlos  ", " Marcos ", "Ana Paula  ", " Lucas "], 
# gere uma lista onde cada nome esteja limpo usando .strip()

lista_operadores = ["  Carlos  ", " Marcos ", "Ana Paula  ", " Lucas "]
lista_limpa = [ o.strip() for o in lista_operadores]

print(lista_limpa)
```

```python
# A7. Dada a lista de status de chamados status_itens = ["concluido", "aberto", "concluido", "aberto", "pendente"], 
# extraia uma lista apenas com os itens cujo texto seja estritamente "aberto"

status_itens = ["concluido", "aberto", "concluido", "aberto", "pendente"]
status_aberto = [status for status in status_itens if status == 'aberto']

print(status_aberto)
```

```python
# A8. Dada a lista de correntes elétricas de bombas em Amperes correntes = [5.2, 6.1, 7.8, 4.9, 9.2, 11.0], 
# crie uma nova lista apenas com as correntes que operaram na faixa nominal segura (entre 5.0 e 8.0 Amperes, inclusive).

correntes = [5.2, 6.1, 7.8, 4.9, 9.2, 11.0]
correntes_nominais = [i for i in correntes if  5.0 <= i <= 8.0]

print(correntes_nominais)
```

```python
#A9. Crie uma lista contendo os múltiplos de 15 entre 1 e 150 usando apenas range() 
# e list comprehension com filtro condicional if

multiplos_15 = [n for n in range(0,151) if n % 15 == 0]
print(multiplos_15)
```

```python
#A10 registros_brutos = ["  TAG-01:OK ", "TAG-02:FALHA  ", " tag-03:ok", "TAG-04:FALHA", " TAG-05:OK "]
# gere uma lista contendo apenas os nomes das TAGs que estão em estado de falha
# com a tag limpa de espaços e padronizada em maiúsculas (saída esperada: ['TAG-02', 'TAG-04']).

registros_brutos = ["  TAG-01:OK ", "TAG-02:FALHA  ", " tag-03:ok", "TAG-04:FALHA", " TAG-05:OK "]

registros_limpos = [
    item.upper().strip().split(":")[0] 
    for item in registros_brutos 
    if item.upper().strip().split(":")[1] == "FALHA"
    ]

print(registros_limpos)

# a estrutura basica de list comprehension é lista_saida = [ expressão(n + 1) | loop (for n in lista_entrada) | condicao (if n == alguma coisa)]

```
