```python
# C2. Crie a função calcular_taxa_refugo(boas, refugadas) que calcula $\frac{refugadas}{boas + refugadas}$. 
# Trate com ZeroDivisionError para retornar 0.0 se a soma de peças produzidas for zero.   

def calcular_taxa_refugo(boas, refugadas):
    try:    # try e exept dentro da função a tornam reutilizaveu em outro contexto 
        taxa = refugadas/(boas+refugadas)
        return taxa
    except ZeroDivisionError:
        return 0.0

resultado_taxa = calcular_taxa_refugo(50, 10)
print(resultado_taxa)
```

```python
# C4. Crie a função obter_sensor(dicionario_leitura, chave) que tenta acessar o valor de uma chave. 
# Trate com KeyError para retornar None caso o sensor solicitado não exista no dicionário.

leituras = {"sensor_01": 50.0}

def obter_sensor(dict_leitura, chave):
    try:
        return dict_leitura[chave]
    except KeyError:
        return None

print(obter_sensor(leituras, "sensor_01"))
print(obter_sensor(leituras, "sensor_02"))
    
```

```python
# C5. Crie a função converter_temperaturas(lista_textos) que recebe uma lista com strings diversas 
# (ex: ["25.4", "OK", "32.1", "ERRO", "", "28.9"]) e retorna uma nova lista contendo apenas os números 
# convertidos para float, ignorando os itens com erro.

lista_str = ["25.4", "OK", "32.1", "ERRO", "", "28.9"]

def converter_temperaturas(lista_textos):
    lista_limpa = []
    
    for item in lista_textos:
        try:
            lista_limpa.append(float(item))
        except ValueError:
            pass

    return lista_limpa
        
print(converter_temperaturas(lista_str))
```

```python
# C7. Escreva a função calcular_media_turnos(valores) que calcula a média aritmética de uma lista. 
# Proteja o código tratando conjuntamente em um único bloco except (ZeroDivisionError, TypeError) 
# para devolver 0.0 caso a lista esteja vazia ou contenha tipos incompatíveis

def calcular_media_turnos(valores_entrada):
    try:
        media_valores = sum(valores_entrada)/ len(valores_entrada)
        return media_valores
    except ZeroDivisionError:
        return 0.0
    except TypeError:
        return 0.0 

valores = [
    (5, 8, 6, 8, 3 , 0),
    ('a',6, 4, 6, 8, 0),
    (0, 0, 0, 0),
    ()
]

medias = [calcular_media_turnos(turno) for turno in valores]
print(medias)




    
```

```python
# C10. Parser de Telemetria defensivo. 

def parsear_telemetria(string_pacote):
    try:
        # 01. Quebra a string por ";"
        str_quebra = string_pacote.split(";")

        # 02. validação existencia de 4 elementos
        if len(str_quebra) != 4: 
            raise IndexError("O pacote precisa ter exatamente 04 elementos")
        
        # 03. conversão de TEMP e PRESSAO para float, RPM para int
        temp_convertida = float(str_quebra[1])
        pressao_convertida = float(str_quebra[2])
        rpm_convertida = int(str_quebra[3])

        # 04. retorna um dict 
        return {
            "linha": str_quebra[0],
            "temp": temp_convertida,
            "pressao": pressao_convertida,
            "rpm": rpm_convertida
        }

    except (ValueError, IndexError, AttributeError):
        return None

pacote_texto = "LINHA;TEMP;PRESSAO;RPM"
pacote_texto_02 = "L1;75.5;2.3;1800"
pacote_texto_03 = "L1;75.5"
pacote_texto_04 = "L1;ERRO;2.3;1800"

print(parsear_telemetria(pacote_texto))
print(parsear_telemetria(pacote_texto_02))
print(parsear_telemetria(pacote_texto_03))
print(parsear_telemetria(pacote_texto_04))
```
