```python
# B1 refugos = [12, 0, 5, 0, 18, 2, 0], gere uma nova lista usando operador ternário (if/else): 
# se o refugo for maior que zero, marque 'Refugo', caso contrário marque 'Aprovado'.

# estrutura do operador tenario com list comprehension: 
# lista_saida = [(valor_se_verdadeiro) if (condicao) else (valor se falso) ]
# a posição do loop vai depender do objetivo

# transformação: lista_saida = [(valor_se_verdadeiro) if (condicao) else (valor se falso) for item in lista ]
# filtragem: lista_saida = [ (expressao) | for item in lista | if (condicao) ] (estrutura basica de list comprehension)

refugos = [12, 0, 5, 0, 18, 2, 0]
status = ["Refugo" if r > 0 else "Aprovado" for r in refugos]

print(status)
```

```python
# B2. Dada a lista paradas = [45, 120, 15, 80, 200, 30], 
# gere uma lista onde paradas >= 60 sejam classificadas como 'Crítica' e paradas menores como 'Leve'.

paradas = [45, 120, 15, 80, 200, 30]
classificacao_paradas = ["Crítica" if p >= 60 else "Leve" for p in paradas]

print(classificacao_paradas)
```

```python
# B3 dada a lista de dicionarios, Gere uma lista contendo 
# apenas as tags ("tag") dos motores que possuem potência acima de 10 CV.

motores = [
    {"tag": "M-01", "potencia_cv": 15, "critico": True},
    {"tag": "M-02", "potencia_cv": 5, "critico": False},
    {"tag": "M-03", "potencia_cv": 50, "critico": True}
]

motores_alta_potencia = [motor['tag'] for motor in motores if motor['potencia_cv'] > 10]
print(motores_alta_potencia)
```

```python
# B4. Usando a mesma lista motores, gere uma lista contendo 
# apenas as tags dos motores que são marcados como críticos (critico == True).

motores_criticos = [motor['tag'] for motor in motores if motor['critico']]
print(motores_criticos)
```

```python
#B5 (Dict Comprehension). Dada a lista de máquinas maquinas = ["Torno", "Fresa", "Injetora", "Prensa"], 
# construa via dict comprehension um dicionário onde a chave é o nome da máquina e o valor é o status 
# inicial padrão "Operacional"

maquinas = ["Torno", "Fresa", "Injetora", "Prensa"]
dict_maquinas = {
    chave: "Operacional" for chave in maquinas
}

print(dict_maquinas)
```

```python
# B6 (Dict Comprehension com Cálculo). Dada a lista de tensões de alimentação tensoes = [220, 380, 440], crie um dicionário 
# onde a chave é a string descritiva (ex: '220V') e o valor é a tensão de pico aproximada (tensão multiplicada por 1.4142).

tensoes_alimentacao = [220, 380, 440]
dict_pico_tensoes = {
    str(chave)+'V': f'{float(chave*1.4142):.2f}' for chave in tensoes_alimentacao
}

print(dict_pico_tensoes)
```

```python
# B7. codigos_tecnicos = {"TEC_01": "Rafael", "TEC_02": "Carlos", "TEC_03": "Lucas"}, crie um novo dicionário via comprehension que inverta a relação 
# (nome do técnico como chave, código como valor)

codigos_tecnicos = {"TEC_01": "Rafael", "TEC_02": "Carlos", "TEC_03": "Lucas"}
codigos_tecnicos_invertidos = {valor: chave for chave, valor in codigos_tecnicos.items()}

print(codigos_tecnicos_invertidos)


```

```python
# B8. pecas_custo = {"Rolamento": 180.0, "Retentor": 45.0, "Placa CPU": 2400.0, "Fusível": 15.0}
# Gere um novo dicionário apenas com as peças cujo custo seja superior a R$ 100,00


pecas_custo = {"Rolamento": 180.0, "Retentor": 45.0, "Placa CPU": 2400.0, "Fusível": 15.0}
custo_pecas = {chave: valor for chave, valor in pecas_custo.items() if valor > 100}

print(custo_pecas)
```

```python
# B9 dada a matriz matriz_producao = [[120, 115, 130], [98, 105, 102], [140, 138, 145]]
# Escreva uma list comprehension que "achate" (flatten) essa lista de listas em uma 
# única lista contínua com todas as 9 produções horárias.

matriz_producao = [[120, 115, 130], [98, 105, 102], [140, 138, 145]]
lista_unica = [numero for sub_lista in matriz_producao for numero in sub_lista] # loops aninhados

print(lista_unica)
```

```python
# B10 Construa, via Dict Comprehension e em um único comando, 
# um dicionário onde a chave seja o número da OS e o valor seja a classificação de severidade:

#Valor 'ALTA' se o tempo for maior ou igual a 120 minutos.
#Valor 'MEDIA' se o tempo estiver entre 60 e 119 minutos.
#Considere no dicionário apenas as ordens do tipo 'Corretiva'.

ordens_manutencao = [
    {"os": 101, "linha": "Linha 1", "tempo_reparo_min": 140, "tipo": "Corretiva"},
    {"os": 102, "linha": "Linha 2", "tempo_reparo_min": 35, "tipo": "Preventiva"},
    {"os": 103, "linha": "Linha 1", "tempo_reparo_min": 85, "tipo": "Corretiva"},
    {"os": 104, "linha": "Linha 3", "tempo_reparo_min": 25, "tipo": "Preditiva"},
    {"os": 105, "linha": "Linha 2", "tempo_reparo_min": 180, "tipo": "Corretiva"}
]

dict_classificacao = {
    item["os"]: "ALTA" if item["tempo_reparo_min"] >= 120 else "MEDIA" 
    for item in ordens_manutencao if item["tipo"] == "Corretiva"
}

#estrutura: {chave: (tenario) + for item in lista + condicao}

print(dict_classificacao)
```
