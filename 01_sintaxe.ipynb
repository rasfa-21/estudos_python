```python
# D2
n = 1
soma = 0

while n <= 100:
    soma += n # soma = soma + n, adicionando o valor de n à variável soma
    n += 1 # n = n + 1, incrementando o valor de n em 1 a cada iteração

print(soma)
```

```python
# D4
for v in vendas:
    if v > 2000:
        print(v) # imprime a venda maior que 2000
        break # interrompe o loop após encontrar o primeiro valor maior que 2000
```

```python
# E1
valor = float(input("Digite o valor do produto: "))
percentual = float(input("Digite o percentual de desconto: "))

def calcular_desconto(valor, percentual):
    desconto = valor * (percentual / 100)
    valor_com_desconto = valor - desconto
    return valor_com_desconto

print(f"O valor do produto com desconto é: R$ {calcular_desconto(valor, percentual):.2f}")
```

```python
# E2
quantidade = int(input("Digite a quantidade de produtos: "))

def classificar_estoque(quantidade):
    if quantidade < 10:
        return "Estoque baixo"
    elif quantidade <= 50:
        return "Estoque médio"
    else:
        return "Estoque alto"

print(classificar_estoque(quantidade))
```

```python
# E3

def resumo_vendas(vendas):
    total_vendas = sum(vendas)
    max_vendas  = max(vendas)
    media_vendas = total_vendas / len(vendas)
    return total_vendas, max_vendas, media_vendas

print(resumo_vendas(vendas))
```

```python
# Bonus

chamados = [
    ['Fazenda Terra Norte', 1200],
    ['Fazenda Boa Vista', 0],
    ['Fazenda Rio Claro', 850],
    ['Fazenda Terra Norte', 2300],
    ['Fazenda Boa Vista', 450],
  ]

soma_valor = 0
chamados_terra_norte = 0

for item in chamados:

  if item[1] != 0:
    soma_valor += item[1]
    if item[0] == 'Fazenda Terra Norte':
      chamados_terra_norte += 1

print(f'O valor total dos chamados é: {soma_valor}')
print(f'O número de chamados na Fazenda Terra Norte é: {chamados_terra_norte}')
```
